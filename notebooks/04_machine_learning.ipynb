# 🔮 Notebook 04: Machine Learning Models & Driver Archetype Clustering
**Formula 1 Performance Intelligence Platform**

Supervised regression for Driver Performance Score, Supervised classification for Top 3 Podium Probability, and K-Means clustering for driver career archetypes.

```python
import sys
import pandas as pd
import numpy as np

sys.path.append('..')
from src.data.loader import F1DataLoader
from src.data.preprocessor import F1DataPreprocessor
from src.features.metrics import F1MetricsCalculator
from src.models.driver_ranking import DriverRankingModel
from src.models.race_prediction import RaceOutcomePredictor
from src.models.driver_clustering import DriverClusterer

loader = F1DataLoader(data_dir='../data')
data_dict = loader.load_all_data()
preprocessor = F1DataPreprocessor(data_dict)
master_df = preprocessor.build_master_results()
calculator = F1MetricsCalculator(master_df, data_dict.get('driver_standings'))
driver_stats = calculator.compute_driver_career_stats(min_starts=10)
print('Data ready for ML training.')
```

## 1. Train Driver Performance Ranking Model

```python
ranking_model = DriverRankingModel()
metrics = ranking_model.fit(driver_stats)
print('Regression Metrics:', metrics)
ranking_model.feature_importances_
```

## 2. Train Race Podium Predictor Classifier

```python
predictor = RaceOutcomePredictor()
cls_metrics = predictor.fit(master_df)
print('Podium Predictor Classification Metrics:', cls_metrics)
```

## 3. Unsupervised K-Means Driver Clustering

```python
clusterer = DriverClusterer(n_clusters=4)
clustered_df = clusterer.fit_predict(driver_stats)
clusterer.get_cluster_centroids(clustered_df)
```
