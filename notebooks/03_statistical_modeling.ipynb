```python
import sys
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.formula.api import ols
import scipy.stats as stats

sys.path.append('..')
from src.data.loader import F1DataLoader
from src.data.preprocessor import F1DataPreprocessor

loader = F1DataLoader(data_dir='../data')
data_dict = loader.load_all_data()
preprocessor = F1DataPreprocessor(data_dict)
master_df = preprocessor.build_master_results()
print('Data ready for statistical modeling.')
```

## 1. ANOVA: Driver vs Constructor Variance Decomposition
Determine how much variance in finishing position is explained by constructor team vs driver.

```python
model_df = master_df[(master_df['year'] >= 2010) & (master_df['is_dnf'] == 0)].copy()
model = ols('positionOrder ~ C(constructorId) + C(driverId)', data=model_df).fit()
anova_table = sm.stats.anova_lm(model, typ=2)
anova_table['variance_explained_%'] = (anova_table['sum_sq'] / anova_table['sum_sq'].sum()) * 100.0
anova_table
```
